```python
# Laplace Smoothing
# To overcome the zero probability problem, we can use Laplace Smoothing.
# Laplace Smoothing adds 1 to every word count so that even unseen words get a non-zero probability.
# It is also called Add-One Smoothing.


from collections import Counter

text = "I love NLP and I love Python"

words = text.lower().split()

word_count = Counter(words)

tot = len(words)
voc = len(word_count)

print("Total words:", tot)
print("Vocabulary:", voc)

print("\nLaplace Smoothing:")

for word in word_count:
    prob = (word_count[word] + 1) / (tot + voc)
    print(word, ":", prob)
unseen_prob = 1 / (tot + voc)
print("Unseen word:", unseen_prob)
```
